# Pipeline

```python
bronze = os.listdir("./BRONZE/")
bronze.sort()
bronze
```

```python
fams = os.listdir("./BRONZE/ArchiveII_famfold/")
fams.sort()
fams
```

```python
tables = {b[10:]: {fam: 999 for fam in fams} for b in bronze}
tables_id = {b[10:]: {fam: 999 for fam in fams} for b in bronze}
```

# Save OneHot encoded data

```python
EMBEDDING = "one-hot"
for file in bronze:
    for fam in fams:
        data_path = f"./BRONZE/{file}/{fam}/{EMBEDDING}/metrics.csv"
        if os.path.exists(data_path):
            df = pd.read_csv(data_path)
            tables[file[10:]][fam] = df["val_f1"].max()
            tables_id[file[10:]][fam] = df["val_f1"].idxmax()
tables_df = pd.DataFrame(tables)
tables_id_df = pd.DataFrame(tables_id)
```

```python
tables_id_df
```

```python
SAVE_PATH = "SILVER/test_tables"
os.makedirs(SAVE_PATH, exist_ok=True)
tables_df.to_csv(f"{SAVE_PATH}/OneHot_best_epochs_test.csv")
tables_id_df.to_csv(f"{SAVE_PATH}/OneHot_best_epochs_idx.csv")
```

# RiNALMo Encoding

```python
EMBEDDING = "RiNALMo"
for file in bronze:
    for fam in fams:
        data_path = f"./BRONZE/{file}/{fam}/{EMBEDDING}/metrics.csv"
        if os.path.exists(data_path):
            df = pd.read_csv(data_path)
            max_val_f1 = df["val_f1"].max()
            max_idx = df["val_f1"].idxmax()
            tables[file[10:]][fam] = max_val_f1
            tables_id[file[10:]][fam] = max_idx
tables_df = pd.DataFrame(tables)
tables_id_df = pd.DataFrame(tables_id)
```

```python
tables_df
```

```python
SAVE_PATH = "SILVER/test_tables"
os.makedirs(SAVE_PATH, exist_ok=True)
tables_df.to_csv(f"{SAVE_PATH}/RiNALMo_best_epochs_test.csv")
tables_id_df.to_csv(f"{SAVE_PATH}/RiNALMo_best_epochs_idx.csv")
```
