# Pipeline

```python
bronze = os.listdir("./BRONZE/")
bronze.sort()
del bronze[1]  # ref
bronze
```

```python
fams = os.listdir("./BRONZE/ArchiveII_famfold/")
fams.sort()
fams
```

```python
tables = {b[10:]: {fam: 999 for fam in fams} for b in bronze}
```

# Save OneHot encoded data

```python
EMBEDDING = "one-hot"
for file in bronze:
    for fam in fams:
        data_path = f"./BRONZE/over_sampled/{file}/{fam}/{EMBEDDING}/metrics.csv"
        if os.path.exists(data_path):
            df = pd.read_csv(data_path)
            tables[file[10:]][fam] = df["val_f1"].iloc[-1]
tables_df = pd.DataFrame(tables)
```

```python
tables_df
```

```python
SAVE_PATH = "SILVER/test_tables_ov"
os.makedirs(SAVE_PATH, exist_ok=True)
tables_df.to_csv(f"{SAVE_PATH}/OneHot_15_epochs_test.csv")
```

# make ref table

```python
embeddings = ["one-hot", "RiNALMo"]
```

```python
tables = {emb: {fam: 999 for fam in fams} for emb in embeddings}

for EMBEDDING in embeddings:
    for fam in fams:
        data_path = f"./BRONZE/ArchiveII_famfold_ref/{EMBEDDING}/{fam}/metrics.csv"
        if os.path.exists(data_path):
            df = pd.read_csv(data_path)
            tables[EMBEDDING][fam] = df["test_f1"].iloc[-1]
    tables_df = pd.DataFrame(tables)
```

```python
SAVE_PATH = "SILVER/test_tables"
os.makedirs(SAVE_PATH, exist_ok=True)
tables_df.to_csv(f"{SAVE_PATH}/ref_15_epochs.csv")
```
