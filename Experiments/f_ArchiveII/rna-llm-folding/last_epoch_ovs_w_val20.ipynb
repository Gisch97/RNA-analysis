# Example

```python
DATA_ROOT = "./BRONZE/over_sampled_w_valid20/"
df = pd.read_csv(
    DATA_ROOT + "ArchiveII_rnadist_200/telomerase/one-hot/metrics_test.csv"
)
print(df)
```

# Pipeline

```python
bronze = os.listdir(DATA_ROOT)
bronze.sort()
# del bronze[1]  # ref
bronze
```

```python
fams = os.listdir(DATA_ROOT + "ArchiveII_famfold/")
fams.sort()
fams
```

```python
tables = {b[10:]: {fam: 999 for fam in fams} for b in bronze}
```

# Save OneHot encoded data

```python
EMBEDDING = "one-hot"
for file in bronze:
    if file == "ArchiveII_famfold":
        for fam in fams:
            data_path = DATA_ROOT + f"{file}/{fam}/{EMBEDDING}/metrics.csv"
            if os.path.exists(data_path):
                df = pd.read_csv(data_path)
                tables[file[10:]][fam] = df["val_f1"].iloc[-1]
    else:
        for fam in fams:
            data_path = DATA_ROOT + f"{file}/{fam}/{EMBEDDING}/metrics_test.csv"
            if os.path.exists(data_path):
                df = pd.read_csv(data_path)
                tables[file[10:]][fam] = df["test_f1"].iloc[-1]
tables_df = pd.DataFrame(tables)
```

```python
tables_df
```

```python
SAVE_PATH = "SILVER/test_tables_ov"
os.makedirs(SAVE_PATH, exist_ok=True)
tables_df.to_csv(f"{SAVE_PATH}/OneHot_val20_15_epochs_test.csv")
```

```python
embeddings = ["one-hot", "RiNALMo"]
```
