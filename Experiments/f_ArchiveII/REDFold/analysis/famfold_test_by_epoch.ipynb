```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
DATA_PATH = "/home/gkulemeyer/Documents/Repos/RNA-analysis/Experiments/f_ArchiveII/REDFold/SILVER/ArchiveII_fam-fold/f1_by_epoch/test_f1.csv"
df = pd.read_csv(DATA_PATH)
df.sample(2)
```

```python
palette = sns.color_palette("tab10") + sns.color_palette("Set2")

plt.figure(figsize=(8, 5))

g = sns.lineplot(
    data=df,
    x="epoch",
    y="test_f1",
    hue="fam",
    markers="o",
    style="fam",
    dashes=False,
    palette=palette,
    alpha=0.7,
    # linestyle="--",
)

# Mejoras de estética
g.set_title("Performance por familia a lo largo de epochs", fontsize=14, weight="bold")
g.set_xlabel("Epochs", fontsize=12)
g.set_ylabel("F1-score en test", fontsize=12)
g.tick_params(axis="both", which="major", labelsize=10)

# Leyenda afuera y ordenada
plt.legend(
    title="Source",
    bbox_to_anchor=(1.01, 1),
    loc="upper left",
    borderaxespad=0,
    fontsize=10,
    title_fontsize=11,
)

# Cuadrícula suave
g.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
# --- 1. Barplot agrupado por familia y epoch ---
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x="fam",
    y="test_f1",
    hue="epoch",
    palette="viridis",
)

plt.title("F1-score promedio por familia y epoch", fontsize=15, weight="bold")
plt.xlabel("Familia", fontsize=12)
plt.ylabel("test_f1", fontsize=12)
plt.legend(title="Epoch", loc="upper right")
plt.grid()
plt.tight_layout()
plt.show()
```
