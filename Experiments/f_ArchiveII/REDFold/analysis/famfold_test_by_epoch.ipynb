```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
DATA_ROOT = "/home/gkulemeyer/Documents/Repos/RNA-analysis/Experiments/f_ArchiveII/REDFold/BRONZE/f1_by_epoch/"
DATA_PATH = DATA_ROOT + "ArchiveII_famfold/test_f1.csv"
df = pd.read_csv(DATA_PATH)
df.sample(2)
```

```python
# --- 1. Barplot agrupado por familia y epoch ---
def plot_f1_by_family_and_epoch(df, dataset:str):
    plt.figure(figsize=(10, 6))

    sns.barplot(
        data=df,
        x="fam",
        y="test_f1",
        hue="epoch",
        palette="viridis",
    )

    plt.title(f"Dataset: {dataset} || F1-score promedio por familia y epoch ", fontsize=15, weight="bold")
    plt.xlabel("Familia", fontsize=12)
    plt.ylabel("test_f1", fontsize=12)
    plt.ylim(0, 0.65)
    plt.legend(title="Epoch", loc="upper right")
    plt.grid()
    plt.tight_layout()
    plt.show()
plot_f1_by_family_and_epoch(df,'famfold')
```

```python
files = os.listdir(DATA_ROOT)
files.sort()
files
```

```python
for file in files:
    DATA_PATH = DATA_ROOT + f"{file}/test_f1.csv"
    df = pd.read_csv(DATA_PATH)
    plot_f1_by_family_and_epoch(df, file[10:])
```

```python
# --- 1. Barplot agrupado por familia y epoch ---
def ax_f1_by_family_and_epoch(df, ax, dataset:str):

    sns.barplot(
        data=df,
        x="fam",
        y="test_f1",
        hue="epoch",
        palette="viridis",
        ax=ax
    )

    ax.set_title(f"Dataset: {dataset} ", fontsize=15, weight="bold")
    ax.set_xlabel("Familia", fontsize=12)
    ax.set_ylabel("test_f1", fontsize=12)
    ax.set_ylim(0, 0.65)
    ax.legend([])
    ax.grid()
    
fig, ax = plt.subplots(1,1,figsize=(8, 6))
ax_f1_by_family_and_epoch(df,ax, 'famfold')
plt.show()
```

```python
hc_files = ['ArchiveII_hc_100',
            'ArchiveII_hc_200',
            'ArchiveII_hc_400',]
rnad_files= ['ArchiveII_rnadist_100',
            'ArchiveII_rnadist_200',
            'ArchiveII_rnadist_400',]

samples_files = [ 'ArchiveII_samples_100',
                'ArchiveII_samples_200',
                'ArchiveII_samples_400']


fig, axs = plt.subplots(1,3, figsize=(20, 5),sharey=True)
for i, file in enumerate(hc_files):
    DATA_PATH = DATA_ROOT + f"{file}/test_f1.csv"
    df = pd.read_csv(DATA_PATH)
    ax_f1_by_family_and_epoch(df, axs[i], file[10:])
plt.tight_layout()
plt.show()


fig, axs = plt.subplots(1,3, figsize=(20, 5),sharey=True)
for i, file in enumerate(rnad_files):
    DATA_PATH = DATA_ROOT + f"{file}/test_f1.csv"
    df = pd.read_csv(DATA_PATH)
    ax_f1_by_family_and_epoch(df, axs[i], file[10:])
plt.tight_layout()
plt.show()



fig, axs = plt.subplots(1,3, figsize=(20, 5),sharey=True)
for i, file in enumerate(samples_files):
    DATA_PATH = DATA_ROOT + f"{file}/test_f1.csv"
    df = pd.read_csv(DATA_PATH)
    ax_f1_by_family_and_epoch(df, axs[i], file[10:])
plt.tight_layout()
plt.show()
```
