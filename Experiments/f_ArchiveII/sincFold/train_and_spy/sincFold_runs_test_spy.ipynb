```python
import os
import pandas as pd
```

# Save tables

```python
# --- Configuración global ---
BASE_PATHS = {
    "famfold": "fam-sim",
    "rnadist_100": "rnadist/dist_100",
    "rnadist_200": "rnadist/dist_200",
    "rnadist_400": "rnadist/dist_400",
    "hc_100": "herarchical_cluster/dist_100",
    "hc_200": "herarchical_cluster/dist_200",
    "hc_400": "herarchical_cluster/dist_400",
    "samples_100": "samples/dist_100",
    "samples_200": "samples/dist_200",
    "samples_400": "samples/dist_400",
}


SAVE_PATH = "../SILVER/train_and_spy/"
```

```python
fams = os.listdir("fam-sim/1_epoch/")
fams
```

```python
tables = {b: {fam: [] for fam in fams} for b in BASE_PATHS.keys()}
for k, v in BASE_PATHS.items():
    for fam in fams:
        for ep in range(1, 16):
            path = f"{v}/{ep}_epoch/{fam}/test.csv"
            tables[k][fam].append(pd.read_csv(path).f1_post.item())
    tab = pd.DataFrame(tables[k])
    # tab.to_csv(f"{SAVE_PATH}{k}_f1_post_test.csv") # SAVE_FILE
```

#  Save Best Epoch

```python
tables_be = {b: {fam: 999 for fam in fams} for b in BASE_PATHS.keys()}
tables_id = {b: {fam: 999 for fam in fams} for b in BASE_PATHS.keys()}

metrics_path = "../SILVER/train_and_spy/"
SAVE_PATH = "../SILVER/train_and_spy/best_epoch/"
```

```python
for k, v in BASE_PATHS.items():
    data_path = f"{metrics_path}{k}_f1_post_test.csv"
    if os.path.exists(data_path):
        df = pd.read_csv(data_path, index_col=0)
        for fam in fams:
            df_fam = df[fam]
            last_val_f1 = df_fam.max()
            max_idx = df_fam.idxmax()
            tables_be[k][fam] = last_val_f1
            tables_id[k][fam] = max_idx
tables_df = pd.DataFrame(tables_be)
tables_id_df = pd.DataFrame(tables_id)
```
