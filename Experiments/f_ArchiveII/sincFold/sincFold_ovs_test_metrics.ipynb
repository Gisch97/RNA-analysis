# Pipeline

```python
bronze = os.listdir(EXP_PATH)
bronze.sort()
# del bronze[0]  # ref
bronze
```

```python
fams = os.listdir(EXP_PATH + "ArchiveII_famfold/")
fams.sort()
fams
```

```python
tables = {b[10:]: {fam: 999 for fam in fams} for b in bronze}
```

# Save OneHot encoded data

```python
for file in bronze:
    for fam in fams:
        data_path = EXP_PATH + f"{file}/{fam}/test.csv"
        if os.path.exists(data_path):
            df = pd.read_csv(data_path)
            tables[file[10:]][fam] = df["f1_post"].iloc[-1]
tables_df = pd.DataFrame(tables)
```
