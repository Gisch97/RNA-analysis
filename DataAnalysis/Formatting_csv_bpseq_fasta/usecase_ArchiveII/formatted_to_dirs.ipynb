# Generate the directories 
Use the dataset splits for training and testing will be used in REDFOLD.

This Notebook Copies files from a source directory with all dataset previously converted to BPSEQ (train) and FASTA (test) to destination directories according to the dataset split.

Converted files were generated using previous scripts format_from_csv.ipynb anc copied to the source directory.

```python
import os
import shutil
import pandas as pd
from tqdm import tqdm
```

```python
REPO_PATH = "/home/gkulemeyer/Documents/Repos/RNA-analysis/DataAnalysis/"

# INPUTS
BPSEQ_FILES = REPO_PATH + "Formatting_csv_bpseq_fasta/usecase_ArchiveII/bpseqs/"
FASTA_FILES = REPO_PATH + "Formatting_csv_bpseq_fasta/usecase_ArchiveII/fastas/"

PARTITIONS_FOLDER = REPO_PATH + "data/filtered_dataset/data_partitions/valid_frac20/"

OUT_PATH = (
    REPO_PATH + "Formatting_csv_bpseq_fasta/usecase_ArchiveII/splits/valid_frac20/"
)
```

```python
def copy_files(out_path, splits, folds, parts):
    """
    The function makes the following structure:
        splits  --- splits --- folds --- parts

        Example:
        splits  --- famfold -|--- 16s-|--- train
                             |        |--- valid
                             |        |--- test
                             |
                             |--- 23s-|--- train
                                      |--- valid
                                      |--- test
        ...
    """
    for fold in folds:
        dirf = out_path + str(fold) + "/"
        if not os.path.exists(dirf):
            os.makedirs(dirf)
        # print(dirf)

        for part in parts:
            dirp = out_path + str(fold) + "/" + part + "/"
            if not os.path.exists(dirp):
                os.makedirs(dirp)
            # print(dirp)

            seq_ids = list(
                splits.loc[(splits.fold == fold) & (splits.partition == part)].id
            )

            # copy files from source to destination
            # print(len(seq_ids))
            for seq_id in tqdm(seq_ids):
                if part == "test":
                    shutil.copy(FASTA_FILES + seq_id + ".fasta", dirp)
                else:
                    shutil.copy(BPSEQ_FILES + seq_id + ".bpseq", dirp)
```

```python
split_files = os.listdir(PARTITIONS_FOLDER)
splits = {file: pd.read_csv(PARTITIONS_FOLDER + file) for file in split_files}
print(split_files[0])
splits[split_files[0]]
```

```python
folds = list(splits[split_files[0]].fold.unique())
parts = list(splits[split_files[0]].partition.unique())
print(f"folds: {folds}")
print(f"parts: {parts}")
```

```python
for file in split_files:
    print(file)
    out_split = OUT_PATH + file[:-4] + "/"
    if not os.path.exists(out_split):
        os.makedirs(out_split)
    copy_files(out_split, splits[file], folds, parts)
```
