```python
def make_data_partition(data_path):
    """
    Create a DataFrame with RNA families and their partitions.
    This function reads the family fold CSV files for each partition and combines them into a single DataFrame.
    Args:
        data_path (str): Path to the directory containing family fold CSV files.
    Returns:
        pd.DataFrame: A DataFrame containing the RNA families, their partitions, and IDs.
    """

    fams = ["16s", "23s", "5s", "RNaseP", "grp1", "srp", "tRNA", "telomerase", "tmRNA"]
    partitions = ["train", "valid", "test"]
    cols = ["fold", "partition", "id"]  # fam, partition, id

    df = pd.DataFrame(columns=cols)
    for p in partitions:
        for fam in fams:
            df_tmp = pd.DataFrame(columns=cols)
            fold = pd.read_csv(data_path + f"{p}_{fam}.csv")
            df_tmp["id"] = fold["id"]
            if p == "test":
                df_tmp["partition"] = "test"
            else:
                df_tmp["partition"] = "train"
            df_tmp["fold"] = fam
            df = pd.concat([df, df_tmp], ignore_index=True)
    df = df.sort_values(by=["fold", "partition", "id"]).reset_index(drop=True)
    return df
```

```python
def df_test_oversampling(df: "pd.DataFrame", target: int = 100) -> "pd.DataFrame":
    """Return a copy of df padded with its own rows (in order, cyclically)
    until it reaches target rows.
    """
    df_out = df.copy()
    i = 0
    while df_out.shape[0] < target:
        df_out = pd.concat([df_out, df.iloc[[i]]], ignore_index=True)
        i = (i + 1) % len(df)
    return df_out


# Aca hay que hacer que este df de test reemplace la familia en train.
```

```python
def make_data_partition_augmented_test(data_path, target=100):
    """
    Create a DataFrame with RNA families and their partitions.
    This function reads the family fold CSV files for each partition and combines them into a single DataFrame.
    Args:
        data_path (str): Path to the directory containing family fold CSV files.
        target (int): target test data shape.
    Returns:
        pd.DataFrame: A DataFrame containing the RNA families, their partitions, and IDs.
    """

    fams = ["16s", "23s", "5s", "RNaseP", "grp1", "srp", "tRNA", "telomerase", "tmRNA"]
    partitions = ["train", "valid", "test"]
    cols = ["fold", "partition", "id"]  # fam, partition, id

    df = pd.DataFrame(columns=cols)
    for p in partitions:
        for fam in fams:
            df_tmp = pd.DataFrame(columns=cols)
            fold = pd.read_csv(data_path + f"{p}_{fam}.csv")
            df_tmp["id"] = fold["id"]
            df_tmp["fold"] = fam
            if p == "test":
                df_tmp["partition"] = "test"
            else:
                df_tmp["partition"] = "train"
                df_tmp = df_test_oversampling(df_tmp, target)
            df = pd.concat([df, df_tmp], ignore_index=True)
    df = df.sort_values(by=["fold", "partition", "id"]).reset_index(drop=True)
    return df
```

```python
SAVE = False
SAVE_PATH = DATA_ROOT + "data_partitions/no_valid"
save_base_name = "ArchiveII_"

fam_fold = "fam-fold"
hc = "herarchical_cluster/"
samples = "samples/"
rnadist = "rnadist/"


dist = [f"dist_{x}" for x in [100]]
```

```python
# Hierarchical Cluster
for d in dist:
    data_path = os.path.join(DATA_ROOT, hc + d + "/")
    df = make_data_partition_augmented_test(data_path)
    print(data_path)
    if SAVE:
        save_name = save_base_name + "hc_" + d[-3:] + ".csv"
        print(f"Saving{SAVE_PATH} /  {save_name}")
        df.to_csv(os.path.join(SAVE_PATH, save_name), index=False)

# RNA DIST
for d in dist:
    data_path = os.path.join(DATA_ROOT, rnadist + d + "/")
    df = make_data_partition_augmented_test(data_path)
    print(data_path)
    if SAVE:
        save_name = save_base_name + "rnadist_" + d[-3:] + ".csv"
        print(f"Saving{SAVE_PATH} /  {save_name}")
        df.to_csv(os.path.join(SAVE_PATH, save_name), index=False)
# SAMPLES
for d in dist:
    data_path = os.path.join(DATA_ROOT, samples + "n_" + d[-3:] + "/")
    df = make_data_partition_augmented_test(data_path)
    print(data_path)
    if SAVE:
        save_name = save_base_name + "samples_n" + d[-3:] + ".csv"
        print(f"Saving{SAVE_PATH} /  {save_name}")
        df.to_csv(os.path.join(SAVE_PATH, save_name), index=False)
```
