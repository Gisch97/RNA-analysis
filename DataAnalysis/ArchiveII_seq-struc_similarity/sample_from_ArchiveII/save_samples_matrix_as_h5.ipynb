```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import rna_utils as ru



DATA_PATH    = "/home/gkulemeyer/Documents/Repos/RNA-analysis/DataAnalysis/data/"

DIST_PATH    = DATA_PATH + "rnadist_f_all.h5" 
META_PATH    = DATA_PATH + "sources/ArchiveII_with_prob_and_motiv.csv"
META100   = DATA_PATH + "filtered_dataset/ArchiveII_samples_100.csv"
META200   = DATA_PATH + "filtered_dataset/ArchiveII_samples_200.csv"
META400   = DATA_PATH + "filtered_dataset/ArchiveII_samples_400.csv"
dist, meta = ru.load_and_align(corr_path=DIST_PATH, meta_path=META_PATH)
```

```python
def build_matrix(dist, ids):
    # Filtrar la matriz de distancias para incluir solo los IDs en la lista
    filtered_dist = dist.loc[ids, ids]
    assert filtered_dist.shape[0] == len(ids), "La matriz filtrada no tiene el tamaño esperado."
    assert filtered_dist.shape[1] == len(ids), "La matriz filtrada no tiene el tamaño esperado."
    return filtered_dist

def get_ids(meta):
    df = pd.read_csv(meta)
    ids = sorted(
df.loc[df['fold_name'] == '16s']["id"].tolist())
    return ids

def get_filtered_dist(dist, meta):
    ids = get_ids(meta)
    filtered_dist = build_matrix(dist, ids)
    return filtered_dist

def save_h5(df_filtered_dist, save_path, save_key): 
    df_filtered_dist.to_hdf(save_path, key=save_key, mode="w")
    return df_filtered_dist
```

```python
df100 =  get_filtered_dist(dist, META100)
df100
```

```python
thresholds = [100, 200, 400] 
paths = [META100, META200, META400]
for threshold, path in zip(thresholds, paths):
    SAVE_PATH = DATA_PATH + f"/samples_filtered{str(threshold)}.h5"
    SAVE_KEY = f"samples_dist{str(threshold)}"
    df = get_filtered_dist(dist, path)
    save_h5(df, SAVE_PATH, SAVE_KEY)
    
```
