```python
# Generate the directories with the dataset splits for training and testing REDFOLD

# Copy files from a source directory with all dataset previously converted to BPSEQ (train) and FASTA (test)
# to destination directories according to the dataset split
# (converted files were generated using previous scripts csv2dirs*.ipynb anc copied to the source directory)
```

```python
import os
import shutil
import pandas as pd
from tqdm import tqdm
```

```python
#CT_FILES = "/DATA/lncRNA/data/archiveii512_ct/"
BPSEQ_FILES = "/DATA/lncRNA/data/archiveii512_bpseq/"
FASTA_FILES = "/DATA/lncRNA/data/archiveii512_fasta/"
DATA_PATH = "/DATA/lncRNA/data/revisiting/"
```

```python
def copy_files(out_path, splits, folds, parts):

    for fold in folds:
        dirf = out_path + str(fold) + "/"
        if not os.path.exists(dirf): os.makedirs(dirf)
        #print(dirf)

        for part in parts:
            dirp = out_path + str(fold) + "/" + part + "/"
            if not os.path.exists(dirp): os.makedirs(dirp)
            #print(dirp)

            seq_ids = list(splits.loc[(splits.fold_number==fold) & \
                                      (splits.partition==part)].id)

            # copy files from source to destination
            #print(len(seq_ids))
            for seq_id in tqdm(seq_ids):
                if part == "test":
                    shutil.copy(FASTA_FILES + seq_id + ".fasta", dirp)
                else:
                    shutil.copy(BPSEQ_FILES + seq_id + ".bpseq", dirp)
```

```python
parts = ["train", "valid", "test"]
```

```python
aiisplits = pd.read_csv(DATA_PATH + "1_random_kfolds/ArchiveII_splits_random.csv")
#display(aiisplits.head())

folds = [0, 1, 2, 3, 4]

OUT_PATH = "/DATA/lncRNA/redfold/splits2/1_random_kfolds_xx/"
if not os.path.exists(OUT_PATH): os.makedirs(OUT_PATH)

copy_files(OUT_PATH, aiisplits, folds, parts)
```
