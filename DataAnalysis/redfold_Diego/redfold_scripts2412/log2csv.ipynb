```python
import pandas as pd
from glob import glob
```

```python
def log2pd(file):
  df = pd.DataFrame()
  with open(file, "r") as f:
    for line in f:
      if line.startswith(">"):
        label = line[1:-1]
        next_line = f.readline().strip()
        next_next_line = f.readline().strip()
        df = df.append({"id": label, "sequence": next_line, "structure": next_next_line}, ignore_index=True)

  return df
```
